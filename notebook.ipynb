## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

### Cleaning Data

#### memperbaiki tipe data dteday dari string jadi datetime dan mengatasi innacurate value pada humidity, serta mengubah data yang sudah di encoding kembali ke label

```python
for df in (df_day, df_hour):
    df['dteday'] = pd.to_datetime(df['dteday'])

print(df_day['dteday'].dtype, df_hour['dteday'].dtype)
```

```python
df_hour[df_hour.hum == df_hour.hum.min()]
```

```python
df_hour['hum'] = df_hour['hum'].replace(0, np.nan)
median_bulanan = df_hour.groupby(['yr', 'mnth'])['hum'].transform('median')
df_hour['hum'] = df_hour['hum'].fillna(median_bulanan)
```

```python
df_hour[df_hour.hum == df_hour.hum.min()]
```

```python
df_day[df_day.hum == df_day.hum.min()]
```

```python
df_day['hum'] = df_day['hum'].replace(0, np.nan)
median_bulanan = df_day.groupby(['yr', 'mnth'])['hum'].transform('median')
df_day['hum'] = df_day['hum'].fillna(median_bulanan)
```

```python
df_day[df_day.hum == df_day.hum.min()]
```

```python
peta_musim = {1: 'Semi', 2: 'Panas', 3: 'Gugur', 4: 'Dingin'}
peta_tahun = {0: 2011, 1: 2012}
peta_hari = {0: 'Minggu', 1: 'Senin', 2: 'Selasa', 3: 'Rabu', 4: 'Kamis', 5: 'Jumat', 6: 'Sabtu'}
peta_cuaca = {1: 'Cerah', 2: 'Berkabut/Berawan', 3: 'Hujan/Salju Ringan', 4: 'Hujan/Salju Lebat'}

def bersihkan(df):
    df = df.copy()
    df['season_label'] = df['season'].map(peta_musim)
    df['year'] = df['yr'].map(peta_tahun)
    df['weekday_label'] = df['weekday'].map(peta_hari)
    df['weather_label'] = df['weathersit'].map(peta_cuaca)
    # Hari libur = akhir pekan + libur resmi (workingday = 0)
    df['day_type'] = np.where(df['workingday'] == 1, 'Hari Kerja', 'Akhir Pekan & Libur')
    df['temp_c'] = df['temp'] * 41
    df['hum_pct'] = df['hum'] * 100
    df['windspeed_kmh'] = df['windspeed'] * 67
    return df

df_day = bersihkan(df_day)
df_hour = bersihkan(df_hour)

print(df_day['day_type'].value_counts())
df_day[['dteday', 'weekday_label', 'day_type', 'casual', 'registered', 'cnt']].head()
```

**Insight:** (Opsional)
- data dteday telah diubah menjadi tipe datetime
- nilai 0 pada hum telah diganti menjadi nilai median

### Eksplorasi Pertanyaan 2: Volume dan komposisi pengguna casual vs registered

```python
rata = df_day.groupby('day_type')[['casual', 'registered', 'cnt']].mean().round(0)
rata['porsi_casual_%'] = (rata['casual'] / rata['cnt'] * 100).round(1)
rata['porsi_registered_%'] = (rata['registered'] / rata['cnt'] * 100).round(1)
display(rata)

selisih = (rata.loc['Akhir Pekan & Libur', ['casual', 'registered', 'cnt']] /
           rata.loc['Hari Kerja', ['casual', 'registered', 'cnt']] - 1) * 100
print('Selisih hari libur dibanding hari kerja (%):')
print(selisih.round(1))
```

```python
# Kontribusi jenis hari terhadap total tiap tipe pengguna
total_tipe = df_day.groupby('day_type')[['casual', 'registered']].sum()
kontribusi = (total_tipe / total_tipe.sum() * 100).round(1)
kontribusi['jumlah_hari'] = df_day['day_type'].value_counts()
kontribusi['porsi_hari_%'] = (kontribusi['jumlah_hari'] / len(df_day) * 100).round(1)
display(kontribusi)
```

```python
# Rata-rata per hari dalam seminggu & pemisahan libur resmi
urut_hari = ['Senin', 'Selasa', 'Rabu', 'Kamis', 'Jumat', 'Sabtu', 'Minggu']
per_hari = df_day.groupby('weekday_label')[['casual', 'registered', 'cnt']].mean().reindex(urut_hari).round(0)
per_hari['porsi_casual_%'] = (per_hari['casual'] / per_hari['cnt'] * 100).round(1)
display(per_hari)

df_day['jenis_detail'] = np.where(df_day['holiday'] == 1, 'Libur Resmi',
                         np.where(df_day['workingday'] == 1, 'Hari Kerja', 'Akhir Pekan'))
detail = df_day.groupby('jenis_detail')[['casual', 'registered', 'cnt']].mean().round(0)
detail['porsi_casual_%'] = (detail['casual'] / detail['cnt'] * 100).round(1)
detail['jumlah_hari'] = df_day['jenis_detail'].value_counts()
display(detail)
```

**Insight:**
- Rata-rata *casual* naik dari ± 607/hari (hari kerja) menjadi ± 1.371/hari (hari libur), atau **+126%**. *Registered* justru turun dari ± 3.978 menjadi ± 2.959 (**−26%**).

## Analisis Lanjutan (Opsional)

Teknik yang digunakan adalah **Clustering tanpa machine learning** :
1. **Manual grouping segmen waktu** – jam dikelompokkan menjadi Dini Hari, Pagi Sibuk, Siang, Sore Sibuk, dan Malam, lalu dibandingkan menurut tipe pengguna. Ini melengkapi Pertanyaan 1 dengan menjawab *pola jam menurut tipe pengguna*.

```python
# 1. Manual grouping segmen waktu
bins_jam = [-1, 5, 9, 15, 19, 23]
label_jam = ['Dini Hari (00-05)', 'Pagi Sibuk (06-09)', 'Siang (10-15)', 'Sore Sibuk (16-19)', 'Malam (20-23)']
df_hour['segmen_waktu'] = pd.cut(df_hour['hr'], bins=bins_jam, labels=label_jam)

# Jam puncak per tipe pengguna dan jenis hari
puncak = df_hour.pivot_table(index='hr', columns='day_type', values=['casual', 'registered'], aggfunc='mean')
for tipe in ['casual', 'registered']:
    for jenis in ['Hari Kerja', 'Akhir Pekan & Libur']:
        s = puncak[(tipe, jenis)]
        print(f'{tipe:<11} | {jenis:<20}: puncak jam {s.idxmax():02d}:00 ({s.max():.0f} sewa/jam)')

# Porsi tiap segmen terhadap total penyewaan tipe pengguna pada jenis hari tersebut
for jenis in ['Hari Kerja', 'Akhir Pekan & Libur']:
    q = df_hour[df_hour['day_type'] == jenis].groupby('segmen_waktu', observed=True)[['casual', 'registered']].sum()
    print(f'\nKontribusi segmen waktu (%) - {jenis}')
    display((q / q.sum() * 100).round(1))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5), sharey=True)
for ax, jenis in zip(axes, ['Hari Kerja', 'Akhir Pekan & Libur']):
    ax.plot(puncak.index, puncak[('registered', jenis)], marker='o', linewidth=2.5,
            color=warna_tipe['registered'], label='Registered')
    ax.plot(puncak.index, puncak[('casual', jenis)], marker='o', linewidth=2.5,
            color=warna_tipe['casual'], label='Casual')
    jr, jc = puncak[('registered', jenis)].idxmax(), puncak[('casual', jenis)].idxmax()
    ax.annotate(f"{jr:02d}:00 · {puncak[('registered', jenis)].max():.0f}", xy=(jr, puncak[('registered', jenis)].max()),
                xytext=(jr, puncak[('registered', jenis)].max() + 40), ha='center',
                color=warna_tipe['registered'], fontweight='bold')
    ax.annotate(f"{jc:02d}:00 · {puncak[('casual', jenis)].max():.0f}", xy=(jc, puncak[('casual', jenis)].max()),
                xytext=(jc, puncak[('casual', jenis)].max() + 40), ha='center',
                color=warna_tipe['casual'], fontweight='bold')
    ax.set_title(jenis, fontweight='bold')
    ax.set_xlabel('Jam dalam sehari')
    ax.set_xticks(range(0, 24, 2))
    ax.set_ylim(0, 560)
axes[0].set_ylabel('Rata-rata penyewaan per jam')
axes[0].legend(title='Tipe pengguna', loc='upper left')
fig.suptitle('Pola Jam Penyewaan menurut Tipe Pengguna', fontsize=15, fontweight='bold')
sns.despine()
plt.tight_layout()
plt.show()
```

**Insight:**
- **Registered hari kerja** memuncak pukul 17:00 (± 468 sewa/jam) dan 08:00 (± 455), pola komuter. **Casual hari kerja** tidak punya puncak pagi dan hanya naik pelan sampai 17:00 (± 57 sewa/jam), sekitar 1/8 dari *registered* pada jam yang sama.
- Pada **hari libur** keduanya memuncak di tengah hari: *registered* pukul 12:00 (± 240) dan *casual* pukul 14:00 (± 140). Segmen siang (10:00–15:00) menyumbang 52,9% *casual* dan 44,2% *registered* pada hari libur.
- Pada hari kerja, segmen Pagi + Sore Sibuk menyumbang 62,7% penyewaan *registered*, tetapi hanya 42,6% *casual*.

## Conclusion & Recommendation

- **Conclusion pertanyaan 1:** Pola jam penyewaan berbeda menurut jenis hari. - **Conclusion pertanyaan 2:** Rata-rata penyewaan *casual* naik **+126%** pada hari libur (± 607 → ± 1.371 per hari), sedangkan *registered* turun **−26%** (± 3.978 → ± 2.959). Porsi *casual* melonjak dari **13,2%** pada hari kerja menjadi **31,7%** pada hari libur. **Rekomendasi Action Item:**
- **Rebalancing hari kerja:** pindahkan sepeda ke area perkantoran dan transit sebelum 07:30 dan 16:30, karena puncak 08:00 dan 17:00 didorong pengguna *registered*.
- **Rebalancing hari libur:** siapkan sepeda di area wisata/taman sebelum 11:00 untuk menyambut puncak sekitar 12:00–14:00.
- **Konversi casual menjadi member:** jalankan penawaran membership (misalnya trial atau diskon tahun pertama) pada Sabtu–Minggu, terutama pukul 11:00–16:00 ketika porsi *casual* tertinggi, karena pada hari libur 31,7% penyewaan berasal dari *casual*.
- **Perawatan armada:** lakukan servis massal pada hari kerja pukul 10:00–15:00 dan dini hari (00:00–05:00); hindari dini hari pada hari libur karena aktivitasnya lebih tinggi.
- **Evaluasi berkala:** pantau porsi *casual* dan *registered* tiap kuartal untuk menilai efektivitas program konversi member.

```python
hour_main_df = df_hour[['dteday', 'hr', 'weekday', 'holiday', 'workingday',
                        'casual', 'registered', 'cnt']]

# df_day -> total harian (dipakai sebagai pembanding/validasi)
day_total_df = df_day[['dteday', 'casual', 'registered', 'cnt']].rename(columns={
    'casual': 'casual_harian',
    'registered': 'registered_harian',
    'cnt': 'cnt_harian'
})

main_data = pd.merge(
    left=hour_main_df,
    right=day_total_df,
    how="left",
    left_on="dteday",
    right_on="dteday"
)

# Validasi: total per jam harus sama dengan total harian
cek = main_data.groupby('dteday').agg(total_jam=('cnt', 'sum'), total_harian=('cnt_harian', 'first'))
print('Hari dengan total tidak sama :', (cek['total_jam'] != cek['total_harian']).sum())
print('Missing value                :', main_data.isna().sum().sum())

main_data.to_csv("main_data.csv", index=False)
main_data.head()
```
